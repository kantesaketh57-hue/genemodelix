```python
import json
import os
from typing import List, Dict, Any, Optional

# Helper to create a sample employees.json if it doesn't exist
def create_sample_file(filepath: str = 'employees.json'):
    sample_data = [
        {"id": 1, "name": "Alice", "department": "Engineering", "salary": 95000, "experience_years": 5},
        {"id": 2, "name": "Bob", "department": "Engineering", "salary": 120000, "experience_years": 8},
        {"id": 3, "name": "Charlie", "department": "HR", "salary": 60000, "experience_years": 3},
        {"id": 4, "name": "David", "department": "Marketing", "salary": 75000, "experience_years": 4},
        {"id": 5, "name": "Eve", "department": "HR", "salary": 65000, "experience_years": 2}
    ]
    if not os.path.exists(filepath):
        with open(filepath, 'w') as f:
            json.dump(sample_data, f, indent=4)
        print(f"Created sample file: {filepath}")

create_sample_file()
```

```python
def load_and_validate_data(filepath: str) -> List[Dict[str, Any]]:
    """Loads data from a JSON file, cleans and validates its structure."""
    if not os.path.exists(filepath):
        raise FileNotFoundError(f"The file {filepath} does not exist.")

    try:
        with open(filepath, 'r') as f:
            data = json.load(f)
    except json.JSONDecodeError as e:
        raise ValueError(f"Invalid JSON format in {filepath}: {e}")

    if not isinstance(data, list):
        raise ValueError("JSON root must be a list of employee objects.")

    # 'id' is treated as optional and fallback-generated if missing
    required_fields = {"name", "department", "salary", "experience_years"}
    validated_employees = []

    for idx, emp in enumerate(data):
        if not isinstance(emp, dict):
            print(f"Warning: Skipping entry at index {idx} because it is not an object.")
            continue

        missing_fields = required_fields - emp.keys()
        if missing_fields:
            print(f"Warning: Skipping employee {emp.get('name', idx)} due to missing fields: {missing_fields}")
            continue

        # Custom clean and parsing function for salary values
        salary_val = emp["salary"]
        if isinstance(salary_val, str):
            # Strip whitespace and convert formats like '90k' or '90K' to float
            clean_sal = salary_val.strip().lower()
            if clean_sal.endswith('k'):
                try:
                    salary_val = float(clean_sal[:-1]) * 1000
                except ValueError:
                    pass

        # Validate types and inject a fallback ID if not present
        try:
            emp_copy = {
                "id": int(emp.get("id", idx + 1)),
                "name": str(emp["name"]),
                "department": str(emp["department"]),
                "salary": float(salary_val),
                "experience_years": float(emp["experience_years"])
            }
            validated_employees.append(emp_copy)
        except (ValueError, TypeError) as e:
            print(f"Warning: Skipping employee {emp.get('name', idx)} due to type conversion error: {e}")

    return validated_employees
```

```python
def analyze_employee_data(employees: List[Dict[str, Any]]) -> Dict[str, Any]:
    """Computes analytical metrics from the validated employee list."""
    if not employees:
        return {
            "average_salary_per_department": {},
            "highest_paid_employee": None,
            "total_experience_years": 0
        }

    dept_salaries = {}
    highest_paid_emp = employees[0]
    total_experience = 0.0

    for emp in employees:
        # Department salary accumulation
        dept = emp["department"]
        dept_salaries.setdefault(dept, []).append(emp["salary"])

        # Highest-paid calculation
        if emp["salary"] > highest_paid_emp["salary"]:
            highest_paid_emp = emp

        # Total experience calculation
        total_experience += emp["experience_years"]

    # Compute average salaries
    avg_salary_per_dept = {
        dept: round(sum(salaries) / len(salaries), 2)
        for dept, salaries in dept_salaries.items()
    }

    return {
        "average_salary_per_department": avg_salary_per_dept,
        "highest_paid_employee": {
            "id": highest_paid_emp["id"],
            "name": highest_paid_emp["name"],
            "department": highest_paid_emp["department"],
            "salary": highest_paid_emp["salary"]
        },
        "total_experience_years": total_experience
    }
```

```python
def write_summary_report(report_data: Dict[str, Any], output_filepath: str = 'summary_report.json'):
    """Writes the processed statistics to a JSON file."""
    try:
        with open(output_filepath, 'w') as f:
            json.dump(report_data, f, indent=4)
        print(f"Successfully wrote summary report to {output_filepath}")
    except IOError as e:
        print(f"Error writing summary report to {output_filepath}: {e}")

# Execution block
try:
    employees = load_and_validate_data('employees.json')
    report = analyze_employee_data(employees)
    write_summary_report(report)

    # Display the written summary
    print("\n--- Summary Report Content ---")
    print(json.dumps(report, indent=4))
except Exception as e:
    print(f"An error occurred during execution: {e}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Re-run pipeline to capture the newly cleaned and validated data
try:
    employees = load_and_validate_data('employees.json')
    report = analyze_employee_data(employees)
    write_summary_report(report)
except Exception as e:
    print(f"Error processing data: {e}")

# Extract average salary per department data
dept_avg_salary = report.get("average_salary_per_department", {})

if dept_avg_salary:
    # Set plot style and figure size
    sns.set_theme(style="whitegrid")
    fig, ax = plt.subplots(figsize=(10, 6))

    # Prepare data for plotting
    departments = list(dept_avg_salary.keys())
    avg_salaries = list(dept_avg_salary.values())

    # Create bar chart
    barplot = sns.barplot(x=departments, y=avg_salaries, hue=departments, palette="viridis", legend=False, ax=ax)

    # Add labels and title
    plt.title("Average Salary per Department (Updated)", fontsize=16, fontweight="bold", pad=15)
    plt.xlabel("Department", fontsize=12, labelpad=10)
    plt.ylabel("Average Salary ($)", fontsize=12, labelpad=10)

    # Add value labels on top of each bar
    for p in barplot.patches:
        height = p.get_height()
        ax.annotate(f"${height:,.2f}",
                    (p.get_x() + p.get_width() / 2., height),
                    ha='center', va='center',
                    xytext=(0, 9),
                    textcoords='offset points',
                    fontsize=10, fontweight="bold")

    # Improve layout and display plot
    plt.tight_layout()
    plt.show()
else:
    print("No department salary data available to display.")
```

```python
import json

# Load the generated summary report
with open('summary_report.json', 'r') as f:
    report_data = json.load(f)

# Print a formatted summary of the results
print("=========================================")
print("         EMPLOYEE SUMMARY REPORT         ")
print("=========================================")

print("\n[1] Average Salary Per Department:")
for dept, avg_sal in report_data["average_salary_per_department"].items():
    print(f"  - {dept}: ${avg_sal:,.2f}")

print("\n[2] Highest Paid Employee:")
emp = report_data["highest_paid_employee"]
if emp:
    print(f"  - Name: {emp['name']}")
    print(f"  - Department: {emp['department']}")
    print(f"  - Salary: ${emp['salary']:,.2f}")
else:
    print("  - No employee data available.")

print(f"\n[3] Cumulative Experience:")
print(f"  - Total Experience: {report_data['total_experience_years']:.1f} years")
print("=========================================")
```
